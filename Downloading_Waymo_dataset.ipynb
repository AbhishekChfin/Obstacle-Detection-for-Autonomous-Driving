```python
# Downloading images at a timestamp skip of 2 sec to avoid non chaning frames.
import pyarrow.dataset as ds
import pyarrow.parquet as pq
import os
import gc


# Paths


image_path = "gs://waymo_open_dataset_v_2_0_1/training/camera_image"

output_dir = "./waymo_extracted"
os.makedirs(output_dir, exist_ok=True)

image_output = os.path.join(
    output_dir,
    "front_camera_images_2sec.parquet"
)

TARGET_IMAGES = 10000
GAP_MICROS = 2_000_000   # 2 seconds



# Dataset


image_dataset = ds.dataset(
    image_path,
    format="parquet"
)

image_scanner = image_dataset.scanner(
    columns=[
        "key.segment_context_name",
        "key.frame_timestamp_micros",
        "key.camera_name",
        "[CameraImageComponent].image"
    ],
    filter=ds.field("key.camera_name") == 1,
    batch_size=64
)



# Temporal sampling


# Last selected timestamp for each segment
last_timestamp = {}

image_writer = None
total_images = 0

for batch in image_scanner.to_batches():

    selected_indices = []

    segments = batch["key.segment_context_name"]
    timestamps = batch["key.frame_timestamp_micros"]

    for i in range(batch.num_rows):

        segment = segments[i].as_py()
        timestamp = timestamps[i].as_py()

        # First frame from this segment
        if segment not in last_timestamp:
            selected_indices.append(i)
            last_timestamp[segment] = timestamp

        # At least 2 seconds since previous selected frame
        elif timestamp - last_timestamp[segment] >= GAP_MICROS:
            selected_indices.append(i)
            last_timestamp[segment] = timestamp

        # Stop once we have enough
        if total_images + len(selected_indices) >= TARGET_IMAGES:
            break

    # Nothing selected from this batch
    if not selected_indices:
        continue

    # Keep only selected rows
    selected_batch = batch.take(selected_indices)

    # Don't exceed TARGET_IMAGES
    remaining = TARGET_IMAGES - total_images

    if selected_batch.num_rows > remaining:
        selected_batch = selected_batch.slice(0, remaining)

    # Initialize writer
    if image_writer is None:
        image_writer = pq.ParquetWriter(
            image_output,
            selected_batch.schema
        )

    image_writer.write_batch(selected_batch)

    total_images += selected_batch.num_rows

    print(
        f"Images extracted: "
        f"{total_images}/{TARGET_IMAGES}"
    )

    del selected_batch
    del batch
    gc.collect()

    if total_images >= TARGET_IMAGES:
        break



# Close


if image_writer is not None:
    image_writer.close()

print("\nExtraction complete.")
print(f"Images: {total_images}")
print(f"Saved: {image_output}")
```

- `pq.read_table()` will load the full data for all specified columns into RAM.

- PyArrow will read every raw compressed `JPEG` byte blob into system memory and construct an in-memory `pyarrow.Table` .

```python
# create an pyarrow table (This will load all the data into the RAM )
import pyarrow.parquet as pq

img_table = pq.read_table(
    "/content/waymo_extracted/front_camera_images_2sec.parquet",
    columns=[
        "key.segment_context_name",
        "key.frame_timestamp_micros",
        "key.camera_name",
        "[CameraImageComponent].image"
    ]
)

keys = [
    "key.segment_context_name",
    "key.frame_timestamp_micros",
    "key.camera_name"
]
```

```python
# compress image keys data
image_keys = set(zip(
    img_table["key.segment_context_name"].to_pylist(),
    img_table["key.frame_timestamp_micros"].to_pylist(),
    img_table["key.camera_name"].to_pylist()
))

print(len(image_keys))
```

```python
# Create an scanner on box image data
box_dataset = ds.dataset(
    "/content/waymo_extracted/front_camera_boxes.parquet",
    format="parquet"
)

box_scanner = box_dataset.scanner(
    columns=[
        "key.segment_context_name",
        "key.frame_timestamp_micros",
        "key.camera_name",
        "key.camera_object_id",
        "[CameraBoxComponent].box.center.x",
        "[CameraBoxComponent].box.center.y",
        "[CameraBoxComponent].box.size.x",
        "[CameraBoxComponent].box.size.y",
        "[CameraBoxComponent].type",
        "[CameraBoxComponent].difficulty_level.detection"
    ],
    batch_size=16384
)
```

```python
# Map the compressed image key data batch-wise in box scanner, and save the data into colab disk
import pyarrow as pa
import gc

box_writer = None
total_boxes = 0

for batch in box_scanner.to_batches():

    segments = batch["key.segment_context_name"].to_pylist()
    timestamps = batch["key.frame_timestamp_micros"].to_pylist()
    cameras = batch["key.camera_name"].to_pylist()

    # Find rows belonging to our selected images
    indices = [
        i
        for i, key in enumerate(zip(segments, timestamps, cameras))
        if key in image_keys
    ]

    if indices:

        matched_batch = batch.take(pa.array(indices))

        if box_writer is None:
            box_writer = pq.ParquetWriter(
                "front_camera_boxes_matched.parquet",
                matched_batch.schema
            )

        box_writer.write_batch(matched_batch)

        total_boxes += matched_batch.num_rows

        print(f"Matched boxes: {total_boxes:,}")

        del matched_batch

    del batch
    gc.collect()



# Close writer


if box_writer is not None:
    box_writer.close()

print("\nBox extraction complete.")
print(f"Total matched boxes: {total_boxes:,}")
```
